# 批量归一化

**批量归一化** (Batch Normalization,BN)，也叫做批量规范化。

训练深层神经网络是十分困难的，特别是在较短的时间内使他们收敛更加棘手。
本节将介绍*批量规范化*（batch normalization） :cite:`Ioffe.Szegedy.2015`，这是一种流行且有效的技术，可持续加速深层网络的收敛速度。
再结合在 :numref:`sec_resnet`中将介绍的残差块，批量规范化使得研究人员能够训练100层以上的网络。

对于为什么要使用批量归一化，这里不再赘述，具体可见[视频](https://www.bilibili.com/list/1567748478)

## 批量归一化层
对于归一化，其实我们不太陌生，在之前的机器学习算法中，对模型进行预处理时，我们通常要对特征值进行归一化。

而批量归一化一般是利用在卷积层与激活层中间（当然，也可以是其他的神经网络，这里以 CNN 为例），对某个小批次的所有同一类特征进行归一化（这里的同一类特征，也就是小批量中同一输出通道）。
公式如下：
$$
z_i=\frac{x_i-\mu}{\sigma}
$$
在实际实现中，为了防止分母置0，我们通常会再加一个参数 $\epsilon$:
$$
z_i = \frac{x_i-\mu}{\sqrt{\sigma^2+\epsilon}}
$$


通常，我们还会对 $z_i$ 进行进一步变换，让归一化的均值和方差不再只为1和0:
$$
\widetilde{z_i}=\gamma\odot z_i + \beta
$$

其中，$\gamma$ 和 $\beta$ 是模型可以学习的参数，以提高模型的泛化性
